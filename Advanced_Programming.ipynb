```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv('/content/data_1.csv')
df2 = pd.read_csv('/content/data_2.csv')
df3 = pd.read_csv('/content/data_3.csv')
df4 = pd.read_csv('/content/data_4.csv')
```

```python
print(df1.isnull().sum())
print(df1.head())
```

```python
print(df2.isnull().sum())
print(df3.isnull().sum())
print(df4.isnull().sum())
```

```python
df1.fillna({'Quantity': 0, 'Price': df1['Price'].mean(), 'Discount': 0}, inplace=True)
df2.fillna({'Quantity': 0, 'Price': df2['Price'].mean(), 'Discount': 0}, inplace=True)
df3.fillna({'Quantity': 0, 'Price': df3['Price'].mean(), 'Discount': 0}, inplace=True)
df4.fillna({'Quantity': 0, 'Price': df4['Price'].mean(), 'Discount': 0}, inplace=True)
```

```python
def calculate_revenue(df):
    df['Net_Price'] = df['Price'] * (1 - df['Discount'])
    df['Total_Revenue'] = df['Quantity'] * df['Net_Price']
    return df

df1 = calculate_revenue(df1)
df2 = calculate_revenue(df2)
df3 = calculate_revenue(df3)
df4 = calculate_revenue(df4)
```

```python
combined_df = pd.concat([df1, df2, df3, df4], ignore_index=True)
```

```python
summary_df = combined_df.groupby('Product').agg({'Quantity': 'sum', 'Total_Revenue': 'sum'}).reset_index()
```

```python
summary_df.to_csv('/content/summary_results.csv', index=False)
```
